# Sales analysis

Totals per *region*.

```python
import os, random
random.seed(int(os.environ.get('NB_SEED', '1')))
sales = {r: random.randint(10, 99) for r in ['north', 'south', 'east']}
print('loaded', len(sales), 'regions')
```

```python
sorted(sales.items())
```

```python
COLOR = (200, 30, 30) if os.environ.get('NB_SEED', '1') == '1' else (30, 30, 200)
import struct, zlib
from IPython.display import Image, HTML, display
def png(w, h, color):
    raw = b''.join(b'\x00' + bytes(color) * w for _ in range(h))
    def chunk(t, d):
        return struct.pack('>I', len(d)) + t + d + struct.pack('>I', zlib.crc32(t + d))
    return b'\x89PNG\r\n\x1a\n' + chunk(b'IHDR', struct.pack('>IIBBBBB', w, h, 8, 2, 0, 0, 0)) + chunk(b'IDAT', zlib.compress(raw)) + chunk(b'IEND', b'')
display(Image(png(8, 4, COLOR)))
```

```python
HTML('<b>total</b>: %d' % sum(sales.values()))
```
